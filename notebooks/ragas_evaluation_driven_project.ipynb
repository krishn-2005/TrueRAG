```python
from ragas.metrics.collections import (
    Faithfulness,
    AnswerRelevancy,
    ContextPrecision,
    ContextRecall
)

faithfulness = Faithfulness(
    llm=evaluator_llm
)

answer_relevancy = AnswerRelevancy(
    llm=evaluator_llm,
    embeddings=ragas_embeddings
)

context_precision = ContextPrecision(
    llm=evaluator_llm
)

context_recall = ContextRecall(
    llm=evaluator_llm
)
```

```python
from ragas import experiment

@experiment()
async def evaluate_rag(row):

    # 1. Faithfulness
    faithfulness_result = await faithfulness.ascore(
        user_input=row["question"],
        response=row["answer"],
        retrieved_contexts=row["context"],
    )

    # 2. Answer Relevancy
    answer_relevancy_result = await answer_relevancy.ascore(
        user_input=row["question"],
        response=row["answer"],
    )

    # 3. Context Precision
    context_precision_result = await context_precision.ascore(
        user_input=row["question"],
        retrieved_contexts=row["context"],
        reference=row["ground_truth"],
    )

    # 4. Context Recall
    context_recall_result = await context_recall.ascore(
        user_input=row["question"],
        retrieved_contexts=row["context"],
        reference=row["ground_truth"],
    )

    return {
        **row,
        "faithfulness": faithfulness_result.value,
        "answer_relevancy": answer_relevancy_result.value,
        "context_precision": context_precision_result.value,
        "context_recall": context_recall_result.value,
    }
```

```python
import time

start = time.time()

result1 = await evaluate_rag.arun(test_dataset)

print("Total Time:", round(time.time() - start, 2), "seconds")
```

```python
import json
import os

CHECKPOINT_FILE = "/content/drive/MyDrive/RAG_Project/data/ragas_checkpoint.json"

if os.path.exists(CHECKPOINT_FILE):
    with open(CHECKPOINT_FILE, "r", encoding="utf-8") as f:
        checkpoint = json.load(f)

    print("Existing checkpoint loaded ✅")

else:
    checkpoint = {}

    print("New checkpoint created ✅")
```

```python
def save_checkpoint():

    with open(CHECKPOINT_FILE, "w", encoding="utf-8") as f:
        json.dump(
            checkpoint,
            f,
            indent=2,
            ensure_ascii=False
        )
```

```python
import time

metric_names = [
    "faithfulness",
    "answer_relevancy",
    "context_precision",
    "context_recall"
]

for experiment_name, rows in experiments.items():

    print(f"\n{'='*60}")
    print(f"Starting: {experiment_name}")
    print(f"{'='*60}")

    if experiment_name not in checkpoint:
        checkpoint[experiment_name] = {}

    for i, row in enumerate(rows):

        row_id = str(i)

        # Already completed row
        if (
            row_id in checkpoint[experiment_name]
            and all(
                metric in checkpoint[experiment_name][row_id]
                for metric in metric_names
            )
        ):
            print(
                f"{experiment_name} | "
                f"Q{i+1}/{len(rows)} | Already completed ✅"
            )
            continue

        if row_id not in checkpoint[experiment_name]:
            checkpoint[experiment_name][row_id] = {
                "question": row["question"]
            }

        result = checkpoint[experiment_name][row_id]

        print(
            f"\n{experiment_name} | "
            f"Q{i+1}/{len(rows)}"
        )

        # -------------------------
        # 1. Faithfulness
        # -------------------------
        if "faithfulness" not in result:

            start = time.time()

            score = await faithfulness.ascore(
                user_input=row["question"],
                response=row["answer"],
                retrieved_contexts=row["context"],
            )

            result["faithfulness"] = score.value

            save_checkpoint()

            print(
                "Faithfulness:",
                score.value,
                "|",
                round(time.time() - start, 2),
                "sec"
            )


        # -------------------------
        # 2. Answer Relevancy
        # -------------------------
        if "answer_relevancy" not in result:

            start = time.time()

            score = await answer_relevancy.ascore(
                user_input=row["question"],
                response=row["answer"],
            )

            result["answer_relevancy"] = score.value

            save_checkpoint()

            print(
                "Answer Relevancy:",
                score.value,
                "|",
                round(time.time() - start, 2),
                "sec"
            )


        # -------------------------
        # 3. Context Precision
        # -------------------------
        if "context_precision" not in result:

            start = time.time()

            score = await context_precision.ascore(
                user_input=row["question"],
                retrieved_contexts=row["context"],
                reference=row["ground_truth"],
            )

            result["context_precision"] = score.value

            save_checkpoint()

            print(
                "Context Precision:",
                score.value,
                "|",
                round(time.time() - start, 2),
                "sec"
            )


        # -------------------------
        # 4. Context Recall
        # -------------------------
        if "context_recall" not in result:

            start = time.time()

            score = await context_recall.ascore(
                user_input=row["question"],
                retrieved_contexts=row["context"],
                reference=row["ground_truth"],
            )

            result["context_recall"] = score.value

            save_checkpoint()

            print(
                "Context Recall:",
                score.value,
                "|",
                round(time.time() - start, 2),
                "sec"
            )


        print("✅ Row completed")
```
