### 1.7 Build the Generation Chain

The basic RAG chain combines:

1. Retrieved context
2. The user's query
3. A context-grounded prompt
4. The local Qwen3 4B model through Ollama
5. A string output parser


# Building Chain

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnablePassthrough, RunnableParallel, RunnableLambda
from langchain_core.output_parsers import StrOutputParser
```

```python
prompt = PromptTemplate(
    template="""
You are a helpful assistant.

Answer the user's question using only the information provided in the context.

If the answer is not present in the context, say:
"I don't have enough information in the provided context."

Context:
{context}

Question:
{query}

Answer:
""",
    input_variables=["context", "query"]
)
```

```python
# from langchain_groq import ChatGroq
from langchain_ollama import ChatOllama

# llm = ChatGroq(
#   model = "openai/gpt-oss-20b"
# )

llm = ChatOllama(
  model="qwen3:4b-q4_K_M",
  reasoning=False,
    keep_alive="2m"
)
```

```python
parser = StrOutputParser()
```

```python
def format_docs(retrieved_docs):
  context_text = "\n\n".join(doc.page_content for doc in retrieved_docs)
  return context_text
```

```python
parallel_chain = RunnableParallel({
  "context" : retriever | RunnableLambda(format_docs),
  "query" : RunnablePassthrough()
})
```

```python
final_chain = parallel_chain | prompt | llm | parser
```

## Phase 2 — LangGraph Upgrade

The baseline RAG flow is converted into a LangGraph workflow so retrieval, generation, and fallback behavior are represented as explicit graph nodes.


### 2.1 Define the Graph State

The graph state carries the user's question, retrieved context, and generated answer.


```python
from typing import TypedDict
from langgraph.graph import StateGraph
```

```python
class State(TypedDict):
    question : str
    context : list[str]
    answer : str
```

### 2.2 Retrieval and Generation Nodes

The retrieval node fetches relevant documents, while the generation node passes the retrieved context and question to the LLM.


```python
async def retrieve_node(state : State):
  question = state["question"]
  retrieved_docs = await retriever.ainvoke(question)
  contexts = [doc.page_content for doc in retrieved_docs]
  
  return {
    "context" : contexts
  }
```

```python
async def generate_node(state : State):
  
  response = await (prompt | llm | parser).ainvoke({
    "context" : state["context"],
    "query" : state["question"]
  })
  
  return {
    "answer" : response
  }
```

### 2.3 Conditional Context Check and Fallback

Before generation, the workflow checks whether usable context was retrieved.

**Flow:**

`START → Retrieve → Context Check → Generate / Fallback → END`


```python
def check_context(state: State):
    context = state.get("context", [])

    if not context:
        return "no_context"

    if not any(str(item).strip() for item in context):
        return "no_context"

    return "has_context"
```

```python
def fallback(state : State):
  return {
    "answer" : "I don't have enough information in the provided documents."
  }
```

```python
from langgraph.graph import StateGraph, START, END

graph = StateGraph(State)

graph.add_node("retrieve", retrieve_node)
graph.add_node("generate", generate_node)
graph.add_node("fallback", fallback)

graph.add_edge(START, "retrieve")
graph.add_conditional_edges(
  "retrieve",
  check_context,
  {
    "has_context" : "generate",
    "no_context" : "fallback"
  }
)

graph.add_edge("generate", END)
graph.add_edge("fallback", END)

app = graph.compile()
```

```python
from ragas import experiment
from ragas.metrics.collections import (
    Faithfulness,
    AnswerRelevancy,
    ContextPrecision,
    ContextRecall,
)
from ragas.embeddings import HuggingFaceEmbeddings
```

```python
from openai import AsyncOpenAI
from ragas.llms import llm_factory

client = AsyncOpenAI(
    api_key="ollama",
    base_url="http://localhost:11434/v1",
)

evaluator_llm = llm_factory(
    "vicgalle/prometheus-7b-v2.0:latest",
    client=client,
    provider="openai",
)

ragas_embeddings = HuggingFaceEmbeddings(
    model="sentence-transformers/all-MiniLM-L6-v2",
    use_api=False
)
```

```python
faithfulness = Faithfulness(
    llm=evaluator_llm
)

answer_relevancy = AnswerRelevancy(
    llm=evaluator_llm,
    embeddings=ragas_embeddings
)

context_precision = ContextPrecision(
    llm=evaluator_llm
)

context_recall = ContextRecall(
    llm=evaluator_llm
)
```

```python
from ragas import Dataset as RagasDataset

experiments = {
    "emb_a_semantic": all_exp["emb_a_semantic"],
    "emb_b_semantic": all_exp["emb_b_semantic"],
    "emb_a_recursive": all_exp["emb_a_recursive"],
    "emb_b_recursive": all_exp["emb_b_recursive"]
}

ragas_datasets = {}

for name, data in experiments.items():

    dataset = RagasDataset(
        name=f"{name}_dataset",
        backend="local/csv",
        root_dir="."
    )

    for row in data:
        dataset.append(row)

    dataset.save()

    ragas_datasets[name] = dataset
```

```python
@experiment()
async def evaluate_rag(row):

    # 1. Faithfulness
    faithfulness_result = await faithfulness.ascore(
        user_input=row["question"],
        response=row["answer"],
        retrieved_contexts=row["context"],
    )

    # 2. Answer Relevancy
    answer_relevancy_result = await answer_relevancy.ascore(
        user_input=row["question"],
        response=row["answer"],
    )

    # 3. Context Precision
    context_precision_result = await context_precision.ascore(
        user_input=row["question"],
        retrieved_contexts=row["context"],
        reference=row["ground_truth"],
    )

    # 4. Context Recall
    context_recall_result = await context_recall.ascore(
        user_input=row["question"],
        retrieved_contexts=row["context"],
        reference=row["ground_truth"],
    )

    return {
        **row,
        "faithfulness": faithfulness_result.value,
        "answer_relevancy": answer_relevancy_result.value,
        "context_precision": context_precision_result.value,
        "context_recall": context_recall_result.value,
    }
```
